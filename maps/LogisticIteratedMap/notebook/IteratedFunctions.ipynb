# Logistic Equation (Iterated Map):
###
A logistic map maps the unit interval to itself:
\begin{equation}
{X_{n+1}} = {RX_n(1 - X_n)} = {RX_n - RX_n^2}
\end{equation}


The itinerary or orbit would have a list of numbers. X is a single state variable, and r is a population parameter:
\begin{equation}
0 < x < 1
\end{equation}
\begin{equation}
N : {0,1,2,3,4,…}
\end{equation}
\begin{equation}
R : (0,4)
\end{equation}
R at max is 4 before the map blows up.



```python
# Xn+1 = RXn(1 - Xn) = RXn - RXn^2
def logisticEquation(r,x0,ntime):
    output = {}
    x_next=x0
    for i in range(ntime):
        x_next=r*x_next*(1-x_next)
        output[i]=x_next        
    return x_next
```

```python
def logisticEquation_trajectory(r,x0,ntime):
    output = {}
    x_next=x0
    for i in range(ntime):
        x_next=r*x_next*(1-x_next)
        output[i]=x_next        
    return output
```

## Changing R parameter:

```python
r=2.9
traj_list_by_ic=[]
for i in range(2):    
    traj_logistic=logisticEquation_trajectory(r,0.2,50)    
    r+=0.2
    traj_list_by_ic.append(traj_logistic.values())


traj_list_by_ic
```

```python
# Changing R parameter:

import matplotlib.pyplot as plt

# plt.figure()
plt.figure(figsize=(10,4))
plt.title(r"Iterated Map by changing R parameter:", {'fontsize': 12,'fontweight': 'bold'})
plt.xlabel(r"iterations (over time)")
plt.ylabel(r'X (population)')
plt.grid()
for i in range(len(traj_list_by_ic)):
    plt.plot(traj_list_by_ic[i],marker='o', markerfacecolor='blue', mouseover=True ,animated=True , )
    plt.legend()
plt.show()
```

# Sensetive Dependence on Initial Conditions: 
### R is static but initial state is changed: 

```python
r=3.46
x0=0.09
traj_list_by_ic=[]
for i in range(2):    
    traj_logistic=logisticEquation_trajectory(r,x0,50)    
    x0+=0.05
    traj_list_by_ic.append(traj_logistic.values())


traj_list_by_ic
```

```python
# Sensetive Dependence on Initial Conditions: R is static but initial state is changed: 

import matplotlib.pyplot as plt

# plt.figure()
plt.figure(figsize=(10,4))
plt.title(r"Sensetive Dependence on Initial Conditions", {'fontsize': 12,'fontweight': 'bold'})
plt.xlabel(r"iterations (over time)")
plt.ylabel(r'X (population)')
plt.grid()
for i in range(len(traj_list_by_ic)):
    plt.plot(traj_list_by_ic[i],marker='o', markerfacecolor='blue', mouseover=True ,animated=True , )
    plt.legend()
plt.show()
```

## Difference in different Trajectories over iterations; highly sensetive on initial conditions:

```python
x0=0.2
x1=0.200001

traj1=[i for i in logisticEquation_trajectory(3.72,x0,100).values()]
traj2=[i for i in logisticEquation_trajectory(3.72,x1,100).values()]
diff= [abs(traj1[i]-traj2[i]) for i in range(100)]

import matplotlib.pyplot as plt

plt.figure(figsize=(12,12))
plt.title(r"Differences in two iterated maps with very close IC over time", {'fontsize': 12,'fontweight': 'bold'})
plt.xlabel(r"iterations (over time)")
plt.ylabel(r'difference')
plt.grid()
plt.plot(range(100),diff,marker='o', markerfacecolor='blue')
plt.show()
```

```python

import numpy as np 
np.mean(diff)
```

## Iterated Cosine Map

```python
import math
def map_cos(x0,ntime):    
    xnext=x0
    for i in range(ntime):
        xnext=math.cos(math.radians(xnext))
    return xnext
```

```python
import math
math.cos(math.radians(math.cos(math.radians(48))))
```

```python
import math

def map_cos_trajectory(x0,ntime):    
    xnext=x0
    traj=[]
    for i in range(ntime):
        xnext=math.cos(math.radians(xnext))
        traj.append(xnext)
    return traj
```
